```python
script = r'''
"""
Test-set evaluation for all saved best-val LMCCS checkpoints on MSRC-21.

For each saved checkpoint in bestValidationModels/, this script:
  1. Mirrors the model dispatch from mcs/models.py:train()
  2. Loads the saved state_dict (via glob, since saved filenames include
     _gt_mode_<mode>_trMode_<train>_ between TASK and DATASET)
  3. Runs inference on the test split
  4. Records test MSE, KTau, NDCG, MAE in a CSV
"""
import os, sys, csv, glob, traceback
import torch

sys.path.insert(0, "/kaggle/working/mcsnet_code_project")
os.chdir("/kaggle/working/mcsnet_code_project")

from common import logger, set_log
from mcs.evaluate import evaluate
import mcs.models as M
from mcs.models import McsData
import argparse


def make_av(task, dataset, gt_mode, gossip_temp=1.0):
    """Argparse-style namespace mirroring mcs/models.py defaults."""
    return argparse.Namespace(
        logpath="logDir/eval", want_cuda=True, has_cuda=torch.cuda.is_available(),
        RUN_TILL_ES=False, ES=10,
        n_layers=3, conv_type='SAGE',
        gt_mode=gt_mode, mcs_mode='edge', training_mode='mse',
        method_type='order', skip='learnable',
        neuromatch_hidden_dim=10, post_mp_dim=64,
        filters_1=10, filters_2=10, filters_3=10, dropout=0,
        COMBO=0, tensor_neurons=10, transform_dim=16,
        bottle_neck_neurons=10, bins=16, histogram=False,
        GMN_NPROPLAYERS=5, MARGIN=0.1, KRON_LAMBDA=0, CONVEX_KRON_LAMBDA=1.0,
        NOISE_FACTOR=0, LP_LOSS_REG=1.0, TEMP=0.1,
        GOSSIP_TEMP=gossip_temp, NITER=20, NUM_GOSSIP_ITER=15,
        NUM_RUNS=2, BATCH_SIZE=128, LEARNING_RATE=0.001, WEIGHT_DECAY=5e-4,
        FEAT_TYPE="One", CONV="GCN",
        DIR_PATH=".", DATASET_NAME=dataset, TASK=task,
        MAX_SET_SIZE=999,
    )


def compute_dataset_stats(av):
    """Mirror what train() does: scan all splits and find max #nodes/#edges."""
    av.MAX_SET_SIZE = 999
    train_data = McsData(av, mode='train')
    max_nodes = max(train_data.query_graph_size_list + train_data.corpus_graph_size_list)
    max_edges = 0
    for g in train_data.query_graphs + train_data.corpus_graphs:
        max_edges = max(max_edges, g.number_of_edges())
    for mode in ('val', 'test'):
        d = McsData(av, mode=mode)
        max_nodes = max(max_nodes, max(d.query_graph_size_list + d.corpus_graph_size_list))
        for g in d.query_graphs + d.corpus_graphs:
            max_edges = max(max_edges, g.number_of_edges())
    return {'max_num_nodes': max_nodes, 'max_num_edges': max_edges}


def get_default_config():
    return {
        'encoder': {'node_hidden_sizes':[10],'node_feature_dim':1,
                    'edge_hidden_sizes':None,'edge_feature_dim':1},
        'aggregator': {'node_hidden_sizes':[10],'graph_transform_sizes':[10],
                       'input_size':[10],'gated':True,'aggregation_type':'sum'},
        'graph_embedding_net': {'node_state_dim':10,'edge_hidden_sizes':[20],
                                'node_hidden_sizes':[10],'n_prop_layers':5,
                                'share_prop_params':True,'edge_net_init_scale':0.1,
                                'node_update_type':'gru','use_reverse_direction':True,
                                'reverse_dir_param_different':False,'layer_norm':False,
                                'prop_type':'embedding'},
        'graph_matching_net': {'node_state_dim':10,'edge_hidden_sizes':[20],
                               'node_hidden_sizes':[10],'n_prop_layers':5,
                               'share_prop_params':True,'edge_net_init_scale':0.1,
                               'node_update_type':'gru','use_reverse_direction':True,
                               'reverse_dir_param_different':False,'layer_norm':False,
                               'prop_type':'matching','similarity':'dotproduct'},
        'model_type':'embedding',
        'graphsim': {'conv_kernel_size':[10,4,2],'linear_size':[24,16],
                     'gcn_size':[10,10,10],'conv_pool_size':[3,3,2],
                     'conv_out_channels':[2,4,8],'dropout':0},
    }


def build_model(av, config, device):
    """Mirror dispatch in mcs/models.py:train() for our three task families."""
    if av.TASK.startswith("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold"):
        av.MAX_SET_SIZE = av.dataset_stats['max_num_nodes']
        m = M.Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold(av, config, 1).to(device)
        return m, "gmn"
    if av.TASK.startswith("Try2Abaltion_NoThresh_IsoNetGossipVar29ForMcs_GossipVector"):
        av.MAX_SET_SIZE = av.dataset_stats['max_num_nodes']
        m = M.Try2Abaltion_NoThresh_IsoNetGossipVar29ForMcs_GossipVector(av, config, 1).to(device)
        return m, "gmn"
    if av.TASK.startswith("IsoNetVar29ForMcs"):
        av.MAX_SET_SIZE = av.dataset_stats['max_num_nodes']
        m = M.IsoNetVar29ForMcs(av, config, 1).to(device)
        return m, "gmn"
    raise ValueError(f"Unknown TASK family: {av.TASK}")


def find_checkpoint(task, gt_mode, dataset):
    """Glob for the saved checkpoint matching this run.

    Saved filenames include `_gt_mode_<mode>_trMode_<train>_` between
    TASK and DATASET, so a bare TASK + '_' + DATASET will not match.
    We use a glob pattern that allows arbitrary characters between TASK,
    gt_mode, and DATASET segments.
    """
    pattern = f"./bestValidationModels/{task}*gt_mode_{gt_mode}*{dataset}"
    matches = glob.glob(pattern)
    if not matches:
        raise FileNotFoundError(f"No checkpoint matching {pattern}")
    if len(matches) > 1:
        # Pick the canonical (shortest) name when multiple match.
        matches = sorted(matches, key=len)
    return matches[0]


# ----------------------------- Run definitions ----------------------------- #
DATASET = "msrc_21_500qgrlarge"

RUNS = [
    # (task, gt_mode, gossip_temp, group_label, lambda_value)
    # λ sweep — 13 entries that actually have saved checkpoints.
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp1e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 0.1,  "lambda_sweep",          0.1),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp5e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 0.5,  "lambda_sweep",          0.5),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp7e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 0.7,  "lambda_sweep",          0.7),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp8e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 0.8,  "lambda_sweep",          0.8),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp9e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 0.9,  "lambda_sweep",          0.9),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp1e+0_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 1.0,  "lambda_sweep_PaperBest",1.0),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp11e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0", "gossip_qap", 1.1,  "lambda_sweep",          1.1),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp12e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0", "gossip_qap", 1.2,  "lambda_sweep",          1.2),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp13e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0", "gossip_qap", 1.3,  "lambda_sweep",          1.3),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp14e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0", "gossip_qap", 1.4,  "lambda_sweep",          1.4),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp15e-1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0", "gossip_qap", 1.5,  "lambda_sweep",          1.5),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp1e+1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 10.0, "lambda_sweep",          10.0),
    ("Try2_IsoNetGossipVar29ForMcs_GossipVectorLRLThreshold_newrun1_FlagOff_temp5e+1_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",  "gossip_qap", 50.0, "lambda_sweep",          50.0),
    # Note: λ=0.05 (temp5e-2) and λ=5.0 (temp5e+0) failed to save best-val
    # checkpoints under early-stopping patience 10, so they are not listed here.

    # Ablation variants for Experiment 1.
    ("IsoNetVar29ForMcs_newrun1_FlagOff_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0",                                                "gossip_qap", 1.0, "ablation_no_gossip",       None),
    ("Try2Abaltion_NoThresh_IsoNetGossipVar29ForMcs_GossipVector_newrun1_FlagOff_temp1e+0_nodeEmbEdgeScoreAdjMask_td16_lr1e-3_noise0","gossip_qap", 1.0, "ablation_no_noisefilter", None),
]

device = "cuda" if torch.cuda.is_available() else "cpu"
config = get_default_config()
results = []

# Compute dataset stats once and build the test sampler once.
av_seed = make_av(RUNS[0][0], DATASET, RUNS[0][1], RUNS[0][2])
stats = compute_dataset_stats(av_seed)
print(f"\n*** Dataset stats: {stats} ***\n")

av_seed.MAX_SET_SIZE = stats['max_num_nodes']
test_data = McsData(av_seed, mode='test')
print(f"*** Test queries: {len(test_data.query_graphs)},  corpus: {len(test_data.corpus_graphs)} ***\n")

for (task, gt, gtemp, group, lam) in RUNS:
    print(f"\n>>> {group}  lambda={lam}")
    try:
        av = make_av(task, DATASET, gt, gtemp)
        av.dataset_stats = stats
        av.MAX_SET_SIZE = stats['max_num_nodes']

        model, dtype = build_model(av, config, device)
        ckpt_path = find_checkpoint(task, gt, DATASET)
        print(f"    loading: {os.path.basename(ckpt_path)}")
        ckpt = torch.load(ckpt_path, weights_only=False)
        model.load_state_dict(ckpt['model_state_dict'])
        model.eval()

        # Re-bind test_data to the current av for gt_mode selection.
        test_data.av = av
        test_data.gt_mode = gt
        test_data.data_type = dtype

        with torch.no_grad():
            ndcg, mse, rankcorr, mae = evaluate(av, model, test_data)
        print(f"    test_mse={mse:.6f}  test_ktau={rankcorr:.6f}  test_ndcg={ndcg:.6f}  test_mae={mae:.6f}")
        results.append({'group': group, 'lambda': lam, 'task': task,
                        'test_mse': mse, 'test_ktau': rankcorr,
                        'test_ndcg': ndcg, 'test_mae': mae, 'error': ''})
    except Exception as e:
        traceback.print_exc()
        results.append({'group': group, 'lambda': lam, 'task': task,
                        'test_mse': None, 'test_ktau': None,
                        'test_ndcg': None, 'test_mae': None,
                        'error': str(e)[:200]})

out_csv = '/kaggle/working/eval_results_msrc.csv'
with open(out_csv, 'w', newline='') as f:
    w = csv.DictWriter(f, fieldnames=list(results[0].keys()))
    w.writeheader()
    w.writerows(results)
print(f"\nSaved: {out_csv}\n")

# Pretty-print summary table
print("=" * 100)
print(f"{'group':<28} {'lambda':>8} {'test_mse':>10} {'test_ktau':>10} {'test_ndcg':>10}")
print("=" * 100)
for r in results:
    if r['error']:
        print(f"{r['group']:<28} {str(r['lambda']):>8} ERROR: {r['error'][:60]}")
    else:
        print(f"{r['group']:<28} {str(r['lambda']):>8} {r['test_mse']:>10.6f} {r['test_ktau']:>10.6f} {r['test_ndcg']:>10.6f}")
'''

with open('/kaggle/working/mcsnet_code_project/eval_all.py', 'w') as f:
    f.write(script)
print("Wrote eval_all.py")
```
